```python
sns.countplot(x='sex', hue='survived', data=df)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='pclass', hue='survived', data=df)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['age'].dropna(), bins=20)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['fare'])

plt.title('Distribution of Passenger Fare')
plt.xlabel('Fare')

plt.show()
```

```python
sns.countplot(x='class', data=df)

plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='class', hue='survived', data=df)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='sex', hue='survived', data=df)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.histplot(data=df, x='age', hue='survived', bins=20, multiple='stack')

plt.title('Survival Distribution by Age')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='embarked', hue='survived', data=df)

plt.title('Survival by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.select_dtypes(include='number').corr(),
            annot=True,
                        cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(df[['survived', 'pclass', 'age', 'fare', 'sibsp', 'parch']].dropna(),
             hue='survived')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='age', y='fare', hue='survived')

plt.title('Age vs Fare by Survival')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

Summary of Findings
The Titanic dataset contains 891 passengers and 15 columns.
342 passengers survived, while 549 passengers did not survive.
Female passengers had a higher number of survivors compared with male passengers.
First-class passengers had better survival outcomes, while third-class passengers had many more non-survivors.
Most passengers were approximately 20–40 years old.
Passenger fares were mostly low, with a few very high fare outliers.
Most passengers embarked from Southampton, where the number of non-survivors was higher.
The heatmap, pairplot, and scatterplot helped identify relationships and patterns among the numerical variables.

Conclusion:
The EDA shows that passenger gender, passenger class, and fare were associated with different survival outcomes. The analysis also revealed the distribution of age, fare, and other passenger characteristics, along with some missing values and outliers.
